```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(42)
```

```python
import matplotlib
print(matplotlib.__version__)
```

```python
x = np.arange(1, 11)
y = x ** 2
plt.plot(x, y)
plt.title("Line Plot")
plt.show()
```

```python
categories = ['AI', 'ML', 'DS', 'Web']
values = [80, 65, 70, 50]
plt.bar(categories, values)
plt.title("Bar Chart")
plt.show()
```

```python
data = np.random.randn(1000)
plt.hist(data, bins=30)
plt.title("Histogram")
plt.show()
```

```python
x = np.random.rand(100)
y = 2 * x + np.random.rand(100)
plt.scatter(x, y)
plt.title("Scatter Plot")
plt.show()
```

```python
data1 = np.random.normal(50, 10, 100)
data2 = np.random.normal(70, 15, 100)
plt.boxplot([data1, data2])
plt.title("Box Plot")
plt.show()
```

```python
labels = ['Python', 'Java', 'C++', 'Others']
sizes = [45, 30, 15, 10]
plt.pie(sizes, labels=labels, autopct='%1.1f%%')
plt.title("Pie Chart")
plt.show()
```

```python
df = pd.DataFrame({
    'Age': np.random.randint(20, 60, 100),
    'Salary': np.random.randint(30000, 100000, 100),
    'Experience': np.random.randint(1, 15, 100)
})
```

```python
sns.heatmap(df.corr(), annot=True)
plt.title("Heatmap")
plt.show()
```

```python
sns.violinplot(data=df)
plt.title("Violin Plot")
plt.show()
```

```python
sns.pairplot(df)
plt.show()
```
